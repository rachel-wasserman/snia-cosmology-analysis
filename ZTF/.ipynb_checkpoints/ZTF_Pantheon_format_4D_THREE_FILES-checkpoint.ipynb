```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from astropy.cosmology import LambdaCDM
from matplotlib.lines import Line2D
from scipy.stats import chi2
import emcee
import corner
from pathlib import Path
```

```python
# Load the three ZTF data tables.
snia = pd.read_csv("ztf_full_data(4).csv")
globalhost = pd.read_csv("globalhost_data(2).csv")
localhost = pd.read_csv("localhost_data(2).csv")

print("SN Ia table:", snia.shape)
print("Global-host table:", globalhost.shape)
print("Local-host table:", localhost.shape)

print("\nSN Ia columns:")
print(snia.columns.tolist())

print("\nGlobal-host columns:")
print(globalhost.columns.tolist())

print("\nLocal-host columns:")
print(localhost.columns.tolist())


def find_shared_key(left, right, candidates):
    for key in candidates:
        if key in left.columns and key in right.columns:
            return key

    shared = [column for column in left.columns if column in right.columns]

    if len(shared) == 1:
        return shared[0]

    raise KeyError(
        "Could not determine a unique merge key. "
        f"Shared columns are: {shared}"
    )


def find_column(data, candidates, description):
    for column in candidates:
        if column in data.columns:
            return column

    raise KeyError(
        f"Could not find {description}. Tried: {candidates}"
    )


# ZTF host-property releases are normally matched by the ZTF transient name.
merge_candidates = [
    "ztfname",
    "ztf_name",
    "ZTFNAME",
    "name",
    "iau_name",
]

global_key = find_shared_key(
    snia,
    globalhost,
    merge_candidates
)

local_key = find_shared_key(
    snia,
    localhost,
    merge_candidates
)

print(f"\nGlobal-host merge key: {global_key}")
print(f"Local-host merge key: {local_key}")


# Keep host columns distinct when similarly named columns exist in both files.
global_rename = {
    column: f"{column}_globalhost"
    for column in globalhost.columns
    if column != global_key and column in snia.columns
}

local_rename = {
    column: f"{column}_localhost"
    for column in localhost.columns
    if column != local_key and column in snia.columns
}

globalhost_merge = globalhost.rename(columns=global_rename)
localhost_merge = localhost.rename(columns=local_rename)

metadata = snia.merge(
    globalhost_merge,
    on=global_key,
    how="left"
)

# If the local table uses a differently named identifier, rename it to the
# global/SN identifier before the second merge.
if local_key != global_key:
    localhost_merge = localhost_merge.rename(columns={local_key: global_key})

metadata = metadata.merge(
    localhost_merge,
    on=global_key,
    how="left"
)


# -------------------------------------------------------------------------
# Identify the columns actually used by the analysis.
# -------------------------------------------------------------------------

redshift_column = find_column(
    metadata,
    [
        "redshift",
        "z",
        "zhel",
    ],
    "ZTF redshift"
)

global_mass_column = find_column(
    metadata,
    [
        "globalmass",
        "global_mass",
        "logmass_global",
        "mass_global",
        "globalmass_globalhost",
    ],
    "global host stellar mass"
)

local_mass_column = find_column(
    metadata,
    [
        "localmass",
        "local_mass",
        "logmass_local",
        "mass_local",
        "localmass_localhost",
    ],
    "local host stellar mass"
)

global_color_column = find_column(
    metadata,
    [
        "globalrestframe_gz",
        "global_restframe_gz",
        "restframe_gz",
        "global_gz",
        "gz",
        "globalrestframe_gz_globalhost",
        "restframe_gz_globalhost",
        "gz_globalhost",
    ],
    "global host rest-frame g-z color"
)

local_color_column = find_column(
    metadata,
    [
        "localrestframe_gz",
        "local_restframe_gz",
        "restframe_gz_local",
        "local_gz",
        "gz_local",
        "localrestframe_gz_localhost",
        "restframe_gz_localhost",
        "gz_localhost",
    ],
    "local host rest-frame g-z color"
)


# Give the merged quantities the common names used throughout this notebook.
metadata["zHD"] = metadata[redshift_column]
metadata["HOST_LOGMASS"] = metadata[global_mass_column]
metadata["LOCAL_LOGMASS"] = metadata[local_mass_column]
metadata["HOST_COLOR"] = metadata[global_color_column]
metadata["LOCAL_COLOR"] = metadata[local_color_column]

print("\nColumns selected for the analysis:")
print(f"redshift          -> {redshift_column}")
print(f"global host mass  -> {global_mass_column}")
print(f"local host mass   -> {local_mass_column}")
print(f"global g-z        -> {global_color_column}")
print(f"local g-z         -> {local_color_column}")


# -------------------------------------------------------------------------
# Standard ZTF sample-quality cuts.
# -------------------------------------------------------------------------

metadata = metadata[metadata["lccoverage_flag"] == 1].copy()
metadata = metadata[metadata["fitquality_flag"] == 1].copy()
metadata = metadata[metadata["fitprob"] >= 1e-7].copy()
metadata = metadata[metadata["redshift"] <= 0.06].copy()
metadata = metadata[
    ~metadata["sn_type"].fillna("").str.contains(
        "pec",
        case=False
    )
].copy()

metadata = metadata[
    (metadata["x0"] > 0)
    & (metadata["zHD"] > 0)
].copy()


# -------------------------------------------------------------------------
# Convert SALT x0 to mB and propagate the SALT covariance matrix.
# -------------------------------------------------------------------------

metadata["mB"] = (
    -2.5 * np.log10(metadata["x0"])
    + 10.5
)

hmBup = (
    -2.5
    * np.log10(metadata["x0"] - metadata["x0_err"])
    + 10.5
)

hmBdn = (
    -2.5
    * np.log10(metadata["x0"] + metadata["x0_err"])
    + 10.5
)

metadata["mB_err"] = (
    hmBup - hmBdn
) / 2.0

dmB_dx0 = (
    -2.5
    / (
        np.log(10.0)
        * metadata["x0"]
    )
)

metadata["COV_mB_x1"] = (
    dmB_dx0
    * metadata["cov_x0_x1"]
)

metadata["COV_mB_c"] = (
    dmB_dx0
    * metadata["cov_x0_c"]
)

metadata["COV_x1_c"] = (
    metadata["cov_x1_c"]
)


# -------------------------------------------------------------------------
# Fixed cosmology for the low-redshift ZTF analysis.
# -------------------------------------------------------------------------

H_0 = 70.0
Omega_M_fixed = 0.3
Omega_Lambda_fixed = 0.7

reference_cosmology = LambdaCDM(
    H0=H_0,
    Om0=Omega_M_fixed,
    Ode0=Omega_Lambda_fixed
)

metadata["MUMODEL"] = (
    reference_cosmology
    .distmod(
        metadata["zHD"].clip(
            lower=1e-6
        )
    )
    .value
)


# -------------------------------------------------------------------------
# Redshift uncertainty.
# -------------------------------------------------------------------------

c_kms = 299792.458
vpec_sigma_kms = 250.0

if "redshift_err" in metadata.columns:
    z_measurement_err = metadata["redshift_err"]
else:
    z_measurement_err = np.zeros(len(metadata))

z_total_err = np.sqrt(
    z_measurement_err**2
    + (
        vpec_sigma_kms
        / c_kms
    )**2
)

metadata["MUERR_VPEC"] = (
    5.0
    / np.log(10.0)
    * z_total_err
    / metadata["zHD"]
)

metadata["LENSDMUERR"] = (
    0.055
    * metadata["zHD"]
)

metadata = metadata.replace(
    [np.inf, -np.inf],
    np.nan
)

print(
    f"\nZTF sample after quality cuts: "
    f"N = {len(metadata)}"
)

print(
    f"Redshift range: "
    f"{metadata['zHD'].min():.5f} "
    f"< z < "
    f"{metadata['zHD'].max():.5f}"
)

print(
    f"Objects with global mass: "
    f"{metadata['HOST_LOGMASS'].notna().sum()}"
)

print(
    f"Objects with local mass: "
    f"{metadata['LOCAL_LOGMASS'].notna().sum()}"
)

print(
    f"Objects with global g-z: "
    f"{metadata['HOST_COLOR'].notna().sum()}"
)

print(
    f"Objects with local g-z: "
    f"{metadata['LOCAL_COLOR'].notna().sum()}"
)
```

## Define sample selections

The ZTF light-curve table is merged with the separate **global-host** and
**local-host** tables before the selections below are defined.

- **mass_cut**: Global host stellar-mass split, using
  $\log_{10}(M_*/M_\odot)=10$.
- **low_z / high_z**: Two ZTF redshift subsets. Since the entire ZTF sample
  is nearby, the visualization split is made at $z=0.04$.
- **low_mass / high_mass**: Global host stellar-mass categories.
- **blue / red**: Global host rest-frame $g-z$ color categories, split at
  the sample median.

The fitted $\gamma$ term uses **global stellar mass**. The separately merged
local mass and local $g-z$ measurements are retained for environmental
comparisons.

```python
mass_cut = 10.0
z_cut = 0.04

low_z = metadata["zHD"] < z_cut
high_z = metadata["zHD"] >= z_cut

low_mass = metadata["HOST_LOGMASS"] < mass_cut
high_mass = metadata["HOST_LOGMASS"] >= mass_cut

color_cut = metadata["HOST_COLOR"].median()
blue = metadata["HOST_COLOR"] < color_cut
red = metadata["HOST_COLOR"] >= color_cut
```

```python
all_samples = {
    "all": metadata,
    }

z_samples = {
    "Low redshift": metadata[low_z],
    "High redshift": metadata[high_z],
    }

z_m_samples = {
    "Low redshift, low-mass hosts": metadata[low_z & low_mass],
    "Low redshift, high-mass hosts": metadata[low_z & high_mass],
    "High redshift, low-mass hosts": metadata[high_z & low_mass],
    "High redshift, high-mass hosts": metadata[high_z & high_mass],
    }
```

## Hubble diagram for the full sample

Plots the ZTF distance modulus versus redshift for every supernova.

At this stage, the distance modulus is constructed using reference SALT standardization values only for visualization. The final Hubble diagram is reconstructed later using the posterior medians from the 4-parameter MCMC fit.

```python
M_B_ref = -19.3
alpha_ref = 0.14
beta_ref = 3.0
gamma_ref = 0.0

HM_ref = (metadata["HOST_LOGMASS"] >= mass_cut).astype(float)

metadata["MU"] = (
    metadata["mB"]
    - M_B_ref
    + alpha_ref * metadata["x1"]
    - beta_ref * metadata["c"]
    + gamma_ref * HM_ref
)

metadata["MUERR"] = np.sqrt(
    metadata["mB_err"]**2
    + (alpha_ref * metadata["x1_err"])**2
    + (beta_ref * metadata["c_err"])**2
    + 2.0 * alpha_ref * metadata["COV_mB_x1"]
    - 2.0 * beta_ref * metadata["COV_mB_c"]
    - 2.0 * alpha_ref * beta_ref * metadata["COV_x1_c"]
    + metadata["MUERR_VPEC"]**2
    + metadata["LENSDMUERR"]**2
)

sample = metadata.dropna(subset=["zHD", "MU", "MUERR", "MUMODEL"]).copy()

plt.figure(figsize=(10, 7))

plt.errorbar(
    sample["zHD"],
    sample["MU"],
    yerr=sample["MUERR"],
    fmt=".",
    markersize=4,
    alpha=0.6,
    capsize=0,
    label="Supernovae"
)

model = sample.sort_values("zHD")

plt.plot(
    model["zHD"],
    model["MUMODEL"],
    linewidth=2,
    label="Reference model"
)

plt.xlabel("Redshift, z")
plt.ylabel(r"Distance modulus, $\mu$")
plt.title(f"Hubble Diagram\nN = {len(sample)}")
plt.grid(alpha=0.2)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
def weighted_mean_and_error(values, errors):
    weights = 1.0 / errors**2
    mean = np.average(values, weights=weights)
    error = np.sqrt(1.0 / weights.sum())
    return mean, error
```

## Hubble residuals versus host galaxy color

This comparison uses the reference-standardized distance moduli above. The final residuals are recalculated later using the posterior medians from the 4D fit.

```python
color_residual_data = metadata.dropna(
    subset=["HOST_COLOR", "MU", "MUERR", "MUMODEL"]
).copy()

color_residual_data["MURES"] = color_residual_data["MU"] - color_residual_data["MUMODEL"]

blue_res = color_residual_data[color_residual_data["HOST_COLOR"] < color_cut]
red_res = color_residual_data[color_residual_data["HOST_COLOR"] >= color_cut]

blue_mean, blue_err = weighted_mean_and_error(blue_res["MURES"], blue_res["MUERR"])
red_mean, red_err = weighted_mean_and_error(red_res["MURES"], red_res["MUERR"])

plt.figure(figsize=(8, 6))
plt.scatter(color_residual_data["HOST_COLOR"], color_residual_data["MURES"], s=15, alpha=0.35)
plt.axhline(0.0, linewidth=1)

plt.hlines(
    blue_mean,
    color_residual_data["HOST_COLOR"].min(),
    color_cut,
    linewidth=3,
    label=f"Bluer mean = {blue_mean:.3f} ± {blue_err:.3f}"
)

plt.hlines(
    red_mean,
    color_cut,
    color_residual_data["HOST_COLOR"].max(),
    linewidth=3,
    label=f"Redder mean = {red_mean:.3f} ± {red_err:.3f}"
)

plt.axvline(color_cut, linestyle="--", linewidth=1)
plt.xlabel("Global host rest-frame g-z")
plt.ylabel("Hubble residual (mag)")
plt.title("Hubble Residuals versus Global Host Color")
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Bluer - redder step = {blue_mean - red_mean:.4f} ± {np.sqrt(blue_err**2 + red_err**2):.4f} mag")
```

### Host Property Correlations

ZTF contains both **global** and **local** host properties. The main cosmology-standardization fit uses global stellar mass for $\gamma$ so that it remains directly comparable to the DES and Pantheon+ analyses. Local mass and local color are treated as secondary environmental diagnostics.

## Cosmological parameter fitting

# Markov Chain Monte Carlo (MCMC) Cosmology Analysis

This notebook performs only:

- **4D:** $M_B$, $\alpha$, $\beta$, $\gamma$
  - full ZTF Hubble-flow sample

The low redshift of the ZTF sample does not provide meaningful leverage to fit $\Omega_M$ and $\Omega_\Lambda$, so they are fixed at

$$
\Omega_M = 0.3, \qquad \Omega_\Lambda = 0.7.
$$

$H_0$ is also fixed at `H_0 = 70.0`. Therefore, the free absolute-magnitude/intercept parameter is $M_B$, not $H_0$.

### Covariance Matrix

The light-curve fitter simultaneously determines three light-curve parameters:

* $x_0$, which is converted into the apparent magnitude $m_B$,
* $x_1$, the stretch parameter,
* $c$, the color parameter.

Because these parameters are fitted simultaneously, their uncertainties are generally correlated. These correlations are described by the covariance matrix

$$
\begin{pmatrix}
\sigma_{m_B}^2 &
\mathrm{Cov}(m_B,x_1) &
\mathrm{Cov}(m_B,c) \\
\mathrm{Cov}(x_1,m_B) &
\sigma_{x_1}^2 &
\mathrm{Cov}(x_1,c) \\
\mathrm{Cov}(c,m_B) &
\mathrm{Cov}(c,x_1) &
\sigma_c^2
\end{pmatrix}.
$$

For the Tripp relation

$$
\mu_{\rm SN}
=
m_B - M_B + \alpha x_1 - \beta c + \gamma H_M,
$$

the light-curve contribution to the variance is

$$
\sigma_\mu^2 =
\sigma_{m_B}^2
+\alpha^2\sigma_{x_1}^2
+\beta^2\sigma_c^2
+2\alpha\,\mathrm{Cov}(m_B,x_1)
-2\beta\,\mathrm{Cov}(m_B,c)
-2\alpha\beta\,\mathrm{Cov}(x_1,c).
$$

A 250 km s$^{-1}$ peculiar-velocity uncertainty and the weak-lensing term are added in quadrature for ZTF.

```python
sigma_int = 0.10

def host_mass_indicator(host_mass):
    return (np.asarray(host_mass) >= mass_cut).astype(float)


def model_distance_modulus(z):
    cosmology = LambdaCDM(
        H0=H_0,
        Om0=Omega_M_fixed,
        Ode0=Omega_Lambda_fixed
    )
    return cosmology.distmod(np.asarray(z)).value


def standardized_mu(data, M_B, alpha, beta, gamma):
    HM = host_mass_indicator(data["HOST_LOGMASS"])
    return (
        data["mB"].to_numpy()
        - M_B
        + alpha * data["x1"].to_numpy()
        - beta * data["c"].to_numpy()
        + gamma * HM
    )


def total_variance(data, alpha, beta):
    variance = (
        data["mB_err"].to_numpy()**2
        + alpha**2 * data["x1_err"].to_numpy()**2
        + beta**2 * data["c_err"].to_numpy()**2
        + 2.0 * alpha * data["COV_mB_x1"].to_numpy()
        - 2.0 * beta * data["COV_mB_c"].to_numpy()
        - 2.0 * alpha * beta * data["COV_x1_c"].to_numpy()
        + data["MUERR_VPEC"].to_numpy()**2
        + data["LENSDMUERR"].to_numpy()**2
        + sigma_int**2
    )

    return variance
```

### 4-parameter model

The fitted parameter vector is

$$
\theta = (M_B, \alpha, \beta, \gamma).
$$

The cosmology is fixed, so the likelihood compares the standardized SN distance modulus directly against the fixed $\Lambda$CDM distance modulus at each ZTF redshift.

```python
def log_prior(theta):
    M_B, alpha, beta, gamma = theta

    if (
        -20.5 < M_B < -18.0
        and 0.0 < alpha < 0.5
        and 0.0 < beta < 6.0
        and -0.5 < gamma < 0.5
    ):
        return 0.0

    return -np.inf


def log_likelihood(theta, data):
    M_B, alpha, beta, gamma = theta

    mu_sn = standardized_mu(data, M_B, alpha, beta, gamma)
    mu_model = model_distance_modulus(data["zHD"])
    variance = total_variance(data, alpha, beta)

    residual = mu_sn - mu_model

    if np.any(~np.isfinite(variance)) or np.any(variance <= 0):
        return -np.inf

    return -0.5 * np.sum(
        np.log(2.0 * np.pi * variance)
        + residual**2 / variance
    )


def log_probability(theta, data):
    lp = log_prior(theta)

    if not np.isfinite(lp):
        return -np.inf

    return lp + log_likelihood(theta, data)
```

```python
def prepare_sample(sample):
    columns = [
        "zHD",
        "mB",
        "mB_err",
        "x1",
        "x1_err",
        "c",
        "c_err",
        "HOST_LOGMASS",
        "COV_mB_x1",
        "COV_mB_c",
        "COV_x1_c",
        "MUERR_VPEC",
        "LENSDMUERR"
    ]

    return (
        sample[columns]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .copy()
    )


def run_mcmc(
    data,
    initial_guess,
    nwalkers=200,
    nsteps=4000,
    burnin=1000,
    thin=10,
    random_seed=123
):
    rng = np.random.default_rng(random_seed)

    ndim = len(initial_guess)
    scales = np.array([0.01, 0.005, 0.05, 0.005])

    starting_positions = (
        initial_guess
        + scales * rng.normal(size=(nwalkers, ndim))
    )

    sampler = emcee.EnsembleSampler(
        nwalkers,
        ndim,
        log_probability,
        args=(data,)
    )

    sampler.run_mcmc(
        starting_positions,
        nsteps,
        progress=True
    )

    samples = sampler.get_chain(
        discard=burnin,
        thin=thin,
        flat=True
    )

    return sampler, samples


def summarize_samples(title, samples, filename=None):
    parameter_names = ["M_B", "alpha", "beta", "gamma"]

    q16, q50, q84 = np.percentile(
        samples,
        [16, 50, 84],
        axis=0
    )

    lines = [title]

    for name, low, median, high in zip(
        parameter_names,
        q16,
        q50,
        q84
    ):
        minus = median - low
        plus = high - median
        line = f"{name} = {median:.5f} -{minus:.5f} +{plus:.5f}"
        print(line)
        lines.append(line)

    if filename is not None:
        with open(filename, "w") as file:
            file.write("\n".join(lines))

    return {
        name: {
            "median": median,
            "minus": median - low,
            "plus": high - median
        }
        for name, low, median, high in zip(
            parameter_names,
            q16,
            q50,
            q84
        )
    }
```

```python
def get_corner_ranges(samples_1, samples_2=None):
    if samples_2 is None:
        combined = samples_1
    else:
        combined = np.vstack([samples_1, samples_2])

    ranges = []

    for i in range(combined.shape[1]):
        low = np.min(combined[:, i])
        high = np.max(combined[:, i])
        padding = 0.05 * (high - low)

        ranges.append((low - padding, high + padding))

    return ranges


def plot_overlaid_corner(
    samples_1,
    samples_2,
    label_1,
    label_2,
    title,
    color_1,
    color_2,
    filename=None
):
    parameter_labels = [
        r"$M_B$",
        r"$\alpha$",
        r"$\beta$",
        r"$\gamma$"
    ]

    corner_ranges = get_corner_ranges(samples_1, samples_2)

    figure = corner.corner(
        samples_1,
        labels=parameter_labels,
        range=corner_ranges,
        color=color_1,
        levels=(0.683, 0.954),
        plot_datapoints=False,
        plot_density=True,
        fill_contours=False,
        smooth=1.0,
        show_titles=False
    )

    corner.corner(
        samples_2,
        fig=figure,
        labels=parameter_labels,
        range=corner_ranges,
        color=color_2,
        levels=(0.683, 0.954),
        plot_datapoints=False,
        plot_density=True,
        fill_contours=False,
        smooth=1.0,
        show_titles=False
    )

    handles = [
        Line2D([0], [0], color=color_1, lw=2, label=label_1),
        Line2D([0], [0], color=color_2, lw=2, label=label_2)
    ]

    figure.legend(handles=handles, loc="upper right", bbox_to_anchor=(0.96, 0.96))
    figure.suptitle(title, fontsize=14, y=1.02)

    if filename is not None:
        figure.savefig(filename, dpi=300, bbox_inches="tight")

    plt.show()
```

```python
def plot_mcmc_diagnostics(
    data,
    samples,
    title,
    filename_prefix,
    clipped=False
):
    M_B, alpha, beta, gamma = np.median(samples, axis=0)

    mu_sn = standardized_mu(data, M_B, alpha, beta, gamma)
    mu_model = model_distance_modulus(data["zHD"])
    residuals = mu_sn - mu_model

    suffix = "_clipped" if clipped else ""

    plt.figure(figsize=(7, 5))
    plt.scatter(data["zHD"], residuals, s=16, alpha=0.5)
    plt.axhline(0.0, linewidth=1)
    plt.xlabel("Redshift, z")
    plt.ylabel("Hubble residual (mag)")
    plt.title(title)
    plt.grid(alpha=0.2)
    plt.tight_layout()
    plt.savefig(
        hubble_dir / f"{filename_prefix}_hubble_residuals{suffix}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()

    plt.figure(figsize=(7, 5))
    plt.hist(residuals, bins=30, alpha=0.7)
    plt.axvline(0.0, linewidth=1)
    plt.xlabel("Hubble residual (mag)")
    plt.ylabel("Number of SNe")
    plt.title(f"{title}: residual distribution")
    plt.tight_layout()
    plt.savefig(
        hubble_dir / f"{filename_prefix}_residual_histogram{suffix}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()

    print(f"{title}:")
    print(f"  N = {len(data)}")
    print(f"  mean residual = {np.mean(residuals):.5f} mag")
    print(f"  residual RMS = {np.sqrt(np.mean(residuals**2)):.5f} mag")


def plot_burnin_diagnostics(
    sampler,
    burnin,
    title,
    filename_prefix,
    clipped=False
):
    labels = [
        r"$M_B$",
        r"$\alpha$",
        r"$\beta$",
        r"$\gamma$"
    ]

    chain = sampler.get_chain()
    suffix = "_clipped" if clipped else ""

    fig, axes = plt.subplots(
        len(labels),
        1,
        figsize=(10, 9),
        sharex=True
    )

    for i, ax in enumerate(axes):
        ax.plot(chain[:, :, i], alpha=0.25)
        ax.axvline(burnin, linestyle="--")
        ax.set_ylabel(labels[i])

    axes[-1].set_xlabel("Step")
    fig.suptitle(f"{title}: walker traces", fontsize=14)

    plt.tight_layout()
    plt.savefig(
        nonclipped_4d_dir / f"{filename_prefix}_burnin{suffix}.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()

    try:
        tau = sampler.get_autocorr_time()

        for label, value in zip(labels, tau):
            print(f"{label}: tau = {value:.1f}")

    except Exception as exc:
        print("Autocorrelation time could not be estimated:")
        print(exc)
```

# Final Hubble Diagrams

```python
M_B = np.median(all_samples_4d[:, 0])
alpha = np.median(all_samples_4d[:, 1])
beta = np.median(all_samples_4d[:, 2])
gamma = np.median(all_samples_4d[:, 3])

data = ztf_sample_4d[
    [
        "zHD",
        "mB",
        "mB_err",
        "x1",
        "x1_err",
        "c",
        "c_err",
        "HOST_LOGMASS",
        "COV_mB_x1",
        "COV_mB_c",
        "COV_x1_c",
        "MUERR_VPEC",
        "LENSDMUERR"
    ]
].dropna().copy()

HM = (data["HOST_LOGMASS"] >= mass_cut).astype(float)

data["MU"] = (
    data["mB"]
    - M_B
    + alpha * data["x1"]
    - beta * data["c"]
    + gamma * HM
)

data["MUERR"] = np.sqrt(total_variance(data, alpha, beta))

cosmology = LambdaCDM(
    H0=H_0,
    Om0=Omega_M_fixed,
    Ode0=Omega_Lambda_fixed
)

z_plot = np.linspace(data["zHD"].min(), data["zHD"].max(), 500)
mu_plot = cosmology.distmod(z_plot).value

plt.figure(figsize=(7, 5))

plt.errorbar(
    data["zHD"],
    data["MU"],
    yerr=data["MUERR"],
    fmt=".",
    markersize=4,
    alpha=0.5,
    capsize=0,
    label="ZTF SNe Ia"
)

plt.plot(
    z_plot,
    mu_plot,
    linewidth=2,
    label="Fixed cosmology"
)

plt.xlabel("Redshift, z")
plt.ylabel(r"Distance modulus, $\mu$")
plt.title(f"ZTF Hubble Diagram\nN = {len(data)}")
plt.grid(alpha=0.2)
plt.legend()

plt.tight_layout()
plt.savefig(
    hubble_dir / "ZTF_final_hubble_diagram_4d.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
data["MUMODEL"] = cosmology.distmod(data["zHD"]).value
data["MURES"] = data["MU"] - data["MUMODEL"]

fig, ax = plt.subplots(figsize=(8, 5))

ax.errorbar(
    data["zHD"],
    data["MURES"],
    yerr=data["MUERR"],
    fmt=".",
    markersize=4,
    alpha=0.5,
    capsize=0
)

ax.axhline(0.0, linewidth=1)
ax.set_xlabel("Redshift, z")
ax.set_ylabel("Hubble residual (mag)")
ax.set_title("ZTF Hubble Residuals")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.savefig(
    hubble_dir / "ZTF_final_hubble_residuals_4d.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

## Global and local host-property residual comparisons

The primary fitted mass correction uses **global host stellar mass**. The cells below compare the final 4D Hubble residuals against global mass, local mass, global color, and local color without changing the standardization model.

```python
final_environment = metadata.copy()

HM = (final_environment["HOST_LOGMASS"] >= mass_cut).astype(float)

final_environment["MU"] = (
    final_environment["mB"]
    - M_B
    + alpha * final_environment["x1"]
    - beta * final_environment["c"]
    + gamma * HM
)

final_environment["MUMODEL"] = cosmology.distmod(
    final_environment["zHD"].clip(lower=1e-6)
).value

final_environment["MURES"] = (
    final_environment["MU"]
    - final_environment["MUMODEL"]
)

global_mass_cut = 10.0
local_mass_cut = final_environment["LOCAL_LOGMASS"].median()
global_color_cut = final_environment["HOST_COLOR"].median()
local_color_cut = final_environment["LOCAL_COLOR"].median()

splits = {
    "Global mass": ("HOST_LOGMASS", global_mass_cut),
    "Local mass": ("LOCAL_LOGMASS", local_mass_cut),
    "Global g-z": ("HOST_COLOR", global_color_cut),
    "Local g-z": ("LOCAL_COLOR", local_color_cut),
}

for name, (column, cut) in splits.items():

    split_data = final_environment[
        [column, "MURES"]
    ].replace([np.inf, -np.inf], np.nan).dropna()

    low = split_data.loc[split_data[column] < cut, "MURES"]
    high = split_data.loc[split_data[column] >= cut, "MURES"]

    step = low.mean() - high.mean()

    step_err = np.sqrt(
        low.var(ddof=1) / len(low)
        + high.var(ddof=1) / len(high)
    )

    print(
        f"{name}: cut = {cut:.3f}, "
        f"N_low = {len(low)}, "
        f"N_high = {len(high)}, "
        f"low - high step = {step:.4f} ± {step_err:.4f} mag"
    )
```
