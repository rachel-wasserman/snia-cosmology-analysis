
# ZTF SN Ia: Fixed-Cosmology SALT2 Analysis

This notebook is the low-redshift ZTF counterpart to the DES and Pantheon+ analyses.

Because the ZTF sample is restricted to very low redshift, the cosmological parameters are not fit. We fix

\[
H_0 = 70\ {\rm km\,s^{-1}\,Mpc^{-1}},\qquad
\Omega_M = 0.3,\qquad
\Omega_\Lambda = 0.7.
\]

The MCMC fit varies only

\[
(M_B,\ \alpha,\ \beta,\ \gamma).
\]

The primary host correction is a **global stellar-mass step** at

\[
\log_{10}(M_*/M_\odot)=10.
\]

Local mass and local color are retained for secondary environmental-split tests.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from astropy.cosmology import LambdaCDM
from pathlib import Path
import emcee
import corner

c_kms = 299792.458
```

## Load ZTF data

```python

def first_existing(*filenames):
    for filename in filenames:
        path = Path(filename)
        if path.exists():
            return path
    raise FileNotFoundError(f"Could not find any of: {filenames}")

ztf_file = first_existing(
    "ztf_full_data(2).csv",
    "ztf_full_data.csv",
    "snia_data(1).csv",
    "snia_data.csv",
)

ztf_raw = pd.read_csv(ztf_file)

print("ZTF file:", ztf_file)
print("ZTF rows:", len(ztf_raw))
print("\nZTF columns:")
print(list(ztf_raw.columns))
```

## Prepare ZTF SALT2 quantities and apply the standard quality cuts

```python

metadata = ztf_raw.copy()

# Standard ZTF DR2-style quality selections.
if "lccoverage_flag" in metadata:
    metadata = metadata[metadata["lccoverage_flag"] == 1].copy()

if "fitquality_flag" in metadata:
    metadata = metadata[metadata["fitquality_flag"] == 1].copy()

if "fitprob" in metadata:
    metadata = metadata[metadata["fitprob"] >= 1e-7].copy()

if "redshift" in metadata:
    metadata = metadata[metadata["redshift"] <= 0.06].copy()

if "sn_type" in metadata:
    metadata = metadata[~metadata["sn_type"].fillna("").str.contains("pec", case=False)].copy()

metadata = metadata[(metadata["x0"] > 0) & (metadata["redshift_HD"] > 0)].copy()

metadata["zHD"] = metadata["redshift_HD"]

# Global properties used in the primary standardization.
metadata["HOST_LOGMASS"] = metadata["globalmass"]
metadata["HOST_COLOR"] = metadata["globalrestframe_gz"]

# Local properties retained for environmental split tests.
metadata["LOCAL_LOGMASS"] = metadata["localmass"]
metadata["LOCAL_COLOR"] = metadata["localrestframe_gz"]

metadata["mB"] = -2.5 * np.log10(metadata["x0"]) + 10.5
metadata["mBERR"] = (2.5 / np.log(10.0)) * metadata["x0_err"] / metadata["x0"]
metadata["x1ERR"] = metadata["x1_err"]
metadata["cERR"] = metadata["c_err"]

dmB_dx0 = -2.5 / (np.log(10.0) * metadata["x0"])
metadata["COV_mB_x1"] = dmB_dx0 * metadata["cov_x0_x1"]
metadata["COV_mB_c"] = dmB_dx0 * metadata["cov_x0_c"]
metadata["COV_x1_c"] = metadata["cov_x1_c"]

print(f"Rows after cuts: {len(metadata)}")
print(f"zHD range: {metadata['zHD'].min():.5f} to {metadata['zHD'].max():.5f}")
```


## Fixed cosmology and nuisance-model choices

The cosmology is held fixed:

\[
H_0=70\ {\rm km\,s^{-1}\,Mpc^{-1}},\qquad
\Omega_M=0.3,\qquad
\Omega_\Lambda=0.7.
\]

The standardized SN distance modulus is

\[
\mu_{\rm SN}
=
m_B-M_B+\alpha x_1-\beta c+\gamma H_M,
\]

where

\[
H_M =
\begin{cases}
0, & \log_{10}(M_*/M_\odot)<10,\\
1, & \log_{10}(M_*/M_\odot)\ge 10.
\end{cases}
\]

The sign convention here means that \(\gamma\) is the additive correction applied to the standardized distance modulus of high-mass hosts.


```python

H0_FIXED = 70.0
OMEGA_M = 0.3
OMEGA_LAMBDA = 0.7

mass_cut = 10.0
hubble_flow_z_min = 0.023

sigma_int = 0.10
vpec_sigma_kms = 250.0

print("Fixed H0 =", H0_FIXED, "km/s/Mpc")
print("Fixed Omega_M =", OMEGA_M)
print("Fixed Omega_Lambda =", OMEGA_LAMBDA)
print("Global host-mass step cut =", mass_cut)
print("Minimum zHD for fit =", hubble_flow_z_min)
```

```python

def salt2_variance(data, alpha, beta, include_redshift=False):
    variance = (
        data["mBERR"].to_numpy()**2
        + alpha**2 * data["x1ERR"].to_numpy()**2
        + beta**2 * data["cERR"].to_numpy()**2
        + 2.0 * alpha * data["COV_mB_x1"].to_numpy()
        - 2.0 * beta * data["COV_mB_c"].to_numpy()
        - 2.0 * alpha * beta * data["COV_x1_c"].to_numpy()
        + sigma_int**2
    )

    if include_redshift:
        z = data["zHD"].to_numpy()
        z_measurement_err = data["redshift_err"].to_numpy()
        z_pec_err = vpec_sigma_kms / c_kms
        z_total_err = np.sqrt(z_measurement_err**2 + z_pec_err**2)

        muerr_redshift = (5.0 / np.log(10.0)) * z_total_err / z
        lensdmuerr = 0.055 * z

        variance = variance + muerr_redshift**2 + lensdmuerr**2

    return variance
```


# Main fit: \(M_B, \alpha, \beta, \gamma\)

The ZTF Hubble-flow sample is fit with a fixed cosmology. The only free parameters are

\[
(M_B,\alpha,\beta,\gamma).
\]

Only SNe with \(z_{\rm HD}>0.023\) are used in the baseline fit to reduce sensitivity to peculiar velocities.


```python

fit_data = metadata[
    [
        "zHD", "redshift_err", "mB", "x1", "c", "HOST_LOGMASS",
        "mBERR", "x1ERR", "cERR",
        "COV_mB_x1", "COV_mB_c", "COV_x1_c"
    ]
].replace([np.inf, -np.inf], np.nan).dropna().copy()

fit_data = fit_data[fit_data["zHD"] > hubble_flow_z_min].copy()

print(f"ZTF Hubble-flow SNe used in fit: {len(fit_data)}")
```

```python

fixed_cosmology = LambdaCDM(
    H0=H0_FIXED,
    Om0=OMEGA_M,
    Ode0=OMEGA_LAMBDA
)


def log_prior(theta):
    M_B, alpha, beta, gamma = theta

    if (
        -20.5 < M_B < -18.0
        and 0.0 < alpha < 0.5
        and 0.0 < beta < 6.0
        and -0.5 < gamma < 0.5
    ):
        return 0.0

    return -np.inf


def log_likelihood(theta):
    M_B, alpha, beta, gamma = theta

    hm = (fit_data["HOST_LOGMASS"].to_numpy() >= mass_cut).astype(float)

    mu_sn = (
        fit_data["mB"].to_numpy()
        - M_B
        + alpha * fit_data["x1"].to_numpy()
        - beta * fit_data["c"].to_numpy()
        + gamma * hm
    )

    mu_model = fixed_cosmology.distmod(fit_data["zHD"].to_numpy()).value
    variance = salt2_variance(fit_data, alpha, beta, include_redshift=True)

    residual = mu_sn - mu_model

    if np.any(~np.isfinite(variance)) or np.any(variance <= 0):
        return -np.inf

    return -0.5 * np.sum(
        np.log(2.0 * np.pi * variance)
        + residual**2 / variance
    )


def log_posterior(theta):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(theta)
```

```python

def run_mcmc(initial_guess, nwalkers=40, nsteps=4000, burnin=1000, thin=5, random_seed=123):
    rng = np.random.default_rng(random_seed)

    scales = np.array([0.01, 0.005, 0.05, 0.005])
    positions = initial_guess + scales * rng.normal(size=(nwalkers, len(initial_guess)))

    for i in range(nwalkers):
        while not np.isfinite(log_prior(positions[i])):
            positions[i] = initial_guess + scales * rng.normal(size=len(initial_guess))

    sampler = emcee.EnsembleSampler(nwalkers, len(initial_guess), log_posterior)
    sampler.run_mcmc(positions, nsteps, progress=True)

    samples = sampler.get_chain(discard=burnin, thin=thin, flat=True)
    return sampler, samples


def summarize_samples(samples):
    labels = ["M_B", "alpha", "beta", "gamma"]
    q16, q50, q84 = np.percentile(samples, [16, 50, 84], axis=0)

    for label, lo, med, hi in zip(labels, q16, q50, q84):
        print(f"{label} = {med:.4f} -{med-lo:.4f} +{hi-med:.4f}")
```

```python

initial_guess = np.array([-19.3, 0.14, 3.0, -0.10])

sampler, samples = run_mcmc(
    initial_guess,
    nwalkers=40,
    nsteps=4000,
    burnin=1000,
    thin=5,
    random_seed=123
)

summarize_samples(samples)

np.save("ZTF_samples_4d.npy", samples)
```

```python

fig = corner.corner(
    samples,
    labels=[r"$M_B$", r"$\\alpha$", r"$\\beta$", r"$\\gamma$"],
    levels=(0.683, 0.954),
    plot_datapoints=False,
    plot_density=True,
    fill_contours=False,
    smooth=1.0,
    show_titles=True,
    title_fmt=".3f"
)

fig.suptitle(
    r"ZTF: fixed $H_0=70$, $\\Omega_M=0.3$, $\\Omega_\\Lambda=0.7$",
    y=1.02
)

plt.show()
```


## Burn-in and chain diagnostics

Do not rely on the corner plot alone. The walker traces should be stationary after burn-in, and the integrated autocorrelation time should be comfortably shorter than the retained chain length.


```python

labels = [r"$M_B$", r"$\\alpha$", r"$\\beta$", r"$\\gamma$"]
chain = sampler.get_chain()

fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

for i, ax in enumerate(axes):
    ax.plot(chain[:, :, i], alpha=0.25)
    ax.axvline(1000, linestyle="--")
    ax.set_ylabel(labels[i])

axes[-1].set_xlabel("Step")
plt.tight_layout()
plt.show()

try:
    tau = sampler.get_autocorr_time()
    for label, value in zip(labels, tau):
        print(f"{label}: tau = {value:.1f}")
except Exception as exc:
    print("Autocorrelation time could not yet be estimated:", exc)
```

## Construct calibrated ZTF distance moduli

```python

M_B_fit, alpha_fit, beta_fit, gamma_fit = np.median(samples, axis=0)

HM = (metadata["HOST_LOGMASS"] >= mass_cut).astype(float)

metadata["MU"] = (
    metadata["mB"]
    - M_B_fit
    + alpha_fit * metadata["x1"]
    - beta_fit * metadata["c"]
    + gamma_fit * HM
)

metadata["MUERR"] = np.sqrt(
    salt2_variance(metadata, alpha_fit, beta_fit, include_redshift=True)
)

metadata["MUMODEL"] = fixed_cosmology.distmod(metadata["zHD"].to_numpy()).value
metadata["MURES"] = metadata["MU"] - metadata["MUMODEL"]

print(f"Fixed H0 = {H0_FIXED:.1f} km/s/Mpc")
print(f"M_B = {M_B_fit:.4f}")
print(f"alpha = {alpha_fit:.4f}")
print(f"beta = {beta_fit:.4f}")
print(f"gamma = {gamma_fit:.4f}")
```

```python

plot_data = metadata.replace([np.inf, -np.inf], np.nan).dropna(
    subset=["zHD", "MU", "MUERR", "MUMODEL"]
).copy()

plt.figure(figsize=(10, 7))
plt.errorbar(
    plot_data["zHD"],
    plot_data["MU"],
    yerr=plot_data["MUERR"],
    fmt=".",
    markersize=4,
    alpha=0.5,
    capsize=0,
    label="ZTF SNe Ia"
)

model = plot_data.sort_values("zHD")
plt.plot(model["zHD"], model["MUMODEL"], linewidth=2, label="Best-fit fixed-density cosmology")

plt.xscale("log")
plt.xlabel("Redshift, z")
plt.ylabel(r"Distance modulus, $\mu$")
plt.title(f"ZTF Hubble Diagram, N = {len(plot_data)}")
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()
```


# Environmental comparisons

The **main \(\gamma\)** is global host mass. Do not replace it with a local quantity in the primary distance-ladder fit unless you also have the equivalent local measurement for the calibrators.

For the ZTF environmental investigation, compare the fitted Hubble residuals using:

1. global stellar mass, with the standard cut at \(\log M_*=10\);
2. local stellar mass, using a median split unless a literature-motivated threshold is desired;
3. global rest-frame \(g-z\) color;
4. local rest-frame \(g-z\) color.

Local color is particularly interesting scientifically for ZTF, but it should be treated as a secondary environmental test here rather than replacing the global-mass correction in the common DES/Pantheon+/ZTF standardization model.


```python

global_mass_cut = 10.0
local_mass_cut = metadata["LOCAL_LOGMASS"].median()
global_color_cut = metadata["HOST_COLOR"].median()
local_color_cut = metadata["LOCAL_COLOR"].median()

splits = {
    "Global mass": ("HOST_LOGMASS", global_mass_cut),
    "Local mass": ("LOCAL_LOGMASS", local_mass_cut),
    "Global g-z": ("HOST_COLOR", global_color_cut),
    "Local g-z": ("LOCAL_COLOR", local_color_cut),
}

for name, (column, cut) in splits.items():
    data = metadata[[column, "MURES"]].replace([np.inf, -np.inf], np.nan).dropna()

    low = data.loc[data[column] < cut, "MURES"]
    high = data.loc[data[column] >= cut, "MURES"]

    step = low.mean() - high.mean()
    step_err = np.sqrt(low.var(ddof=1) / len(low) + high.var(ddof=1) / len(high))

    print(
        f"{name}: cut={cut:.3f}, "
        f"N_low={len(low)}, N_high={len(high)}, "
        f"low-high step={step:.4f} +/- {step_err:.4f} mag"
    )
```
