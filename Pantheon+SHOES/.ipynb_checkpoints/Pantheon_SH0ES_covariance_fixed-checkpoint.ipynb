````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from astropy.cosmology import LambdaCDM
from matplotlib.lines import Line2D
from scipy.stats import chi2
import emcee
import corner
from pathlib import Path
````

````python
metadata = pd.read_csv("Pantheon+SH0ES.dat", sep=r"\s+")
metadata["_PANTHEON_ROW"] = np.arange(len(metadata))

# Full Pantheon+SH0ES statistical + systematic covariance matrix for 2D/3D cosmology fits.
def load_pantheon_covariance(filename):
    values = np.loadtxt(filename).reshape(-1)
    n = int(values[0])
    covariance_values = values[1:]

    if covariance_values.size != n*n:
        raise ValueError(f"Expected {n*n} covariance elements, found {covariance_values.size}")

    return covariance_values.reshape(n, n)

pantheon_cov = load_pantheon_covariance("Pantheon+SH0ES_STAT+SYS.cov")

if pantheon_cov.shape != (len(metadata), len(metadata)):
    raise ValueError(
        f"Covariance shape {pantheon_cov.shape} does not match Pantheon table length {len(metadata)}"
    )


# Match the column names used throughout the original notebook.
metadata["MU"] = metadata["MU_SH0ES"]
# Plotting/error-bar quantity only; 2D/3D fits use pantheon_cov.
metadata["MUERR"] = metadata["MU_SH0ES_ERR_DIAG"]
metadata["MUERR_VPEC"] = metadata["m_b_corr_err_VPEC"]
metadata["LENSDMUERR"] = 0.055 * metadata["zHD"]

# Construct a reference-model distance modulus for plotting/residual figures.
reference_cosmology = LambdaCDM(H0=73.04, Om0=0.3, Ode0=0.7)
metadata["MUMODEL"] = reference_cosmology.distmod(metadata["zHD"].clip(lower=1e-6)).value

# Convert the released x0 covariances to mB covariances, as in the DES notebook.
dmB_dx0 = -2.5 / (np.log(10.0) * metadata["x0"])
metadata["COV_mB_x1"] = dmB_dx0 * metadata["COV_x1_x0"]
metadata["COV_mB_c"] = dmB_dx0 * metadata["COV_c_x0"]

# Sentinel values in the release use -9 or -999 for missing quantities.
metadata = metadata.replace([-9, -999], np.nan)
````

## Define sample selections

- **mass_cut**: Host stellar-mass split, kept at the same value used in the DES notebook.
- **low_z / high_z**: Low- and high-redshift supernovae.
- **low_mass / high_mass**: Host galaxy stellar mass categories.

The public Pantheon+SH0ES distance table does not include the DES notebook's host SFR, sSFR, or host-color columns, so those splits are not reproduced here.


````python
mass_cut = 10.0

low_z = metadata["zHD"] < 0.1
high_z = metadata["zHD"] >= 0.1

low_mass = metadata["HOST_LOGMASS"] < mass_cut
high_mass = metadata["HOST_LOGMASS"] >= mass_cut
````

````python
all_samples = {
    "all": metadata,
    }

z_samples = {
    "Low redshift": metadata[low_z],
    "High redshift": metadata[high_z],
    }

z_m_samples = {
    "Low redshift, low-mass hosts": metadata[low_z & low_mass],
    "Low redshift, high-mass hosts": metadata[low_z & high_mass],
    "High redshift, low-mass hosts": metadata[high_z & low_mass],
    "High redshift, high-mass hosts": metadata[high_z & high_mass],
    }
````

## Cosmological parameter fitting

`MU_SH0ES_ERR_DIAG` is retained only for plotting/error bars. The 2D and 3D cosmology fits use the full `Pantheon+SH0ES_STAT+SYS.cov` covariance matrix. Each sample and host-mass split keeps the original Pantheon+SH0ES row number in `_PANTHEON_ROW`, so the corresponding covariance submatrix is selected correctly. The likelihood uses a Cholesky decomposition to evaluate the correlated Gaussian form without explicitly inverting the covariance matrix.


# Markov Chain Monte Carlo (MCMC) Cosmology Analysis

The goal of the MCMC is to determine the cosmological parameters that best reproduce the observed Type Ia supernova Hubble diagram.

## Step 1: Choose the Cosmological Model

Three cosmological models:

### Flat ΛCDM (2 parameters)

The free parameters are

- $H_0$
- $\Omega_M$

while the dark-energy density is constrained by spatial flatness:

$$
\Omega_\Lambda = 1-\Omega_M.
$$

---

### Non-flat ΛCDM (3 parameters)

The free parameters are

- $H_0$
- $\Omega_M$
- $\Omega_\Lambda$

where all three parameters are allowed to vary independently.

---

### Non-flat ΛCDM + SALT2 Standardization (6 parameters)

The free parameters are

- $H_0$
- $\Omega_M$
- $\Omega_\Lambda$
- $\alpha$
- $\beta$
- $\gamma$

where

- $\alpha$ is the stretch coefficient,
- $\beta$ is the color coefficient,
- $\gamma$ is the host-mass correction.

The observed distance modulus is no longer taken directly from the metadata. Instead, it is reconstructed for every proposed parameter set as

$$
\mu_{\rm obs} = m_B - M_B + \alpha x_1 - \beta c + \gamma HM,
$$

where

- $m_B$ is the apparent peak magnitude,
- $M_B=-19.3$ is the fixed absolute magnitude,
- $x_1$ is the stretch parameter,
- $c$ is the color parameter,
- $HM$ is the host-mass indicator,

defined as

$$
HM =
\begin{cases}
0,&\log(M_*/M_\odot)<10,\\
1,&\log(M_*/M_\odot)\ge10.
\end{cases}
$$

This allows the MCMC to simultaneously determine both the cosmological parameters and the supernova standardization parameters.

---

## Step 2: Load the Supernova Data

For the two- and three-parameter models, each supernova contributes

- redshift, $z$,
- observed distance modulus, $\mu$,
- distance-modulus uncertainty, $\sigma_\mu$.

For the six-parameter model, the raw quantities are used instead:

- redshift, $z$,
- apparent magnitude, $m_B$,
- stretch parameter, $x_1$,
- color parameter, $c$,
- host stellar mass,
- measurement uncertainties.

The observed distance modulus is reconstructed inside the likelihood using the current trial values of $\alpha$, $\beta$, and $\gamma$.

---

## Step 3: Generate Initial Walker Positions

The MCMC begins with an initial parameter guess, for example

$$
H_0 = 70,\qquad
\Omega_M = 0.30,\qquad
\Omega_\Lambda = 0.70.
$$

Rather than starting from a single point, an ensemble of walkers is created by randomly perturbing the initial guess.

Example walker positions:

| Walker | $H_0$ | $\Omega_M$ | $\Omega_\Lambda$ |
|--------|--------:|-------------:|-------------------:|
| 1 | 69.8 | 0.31 | 0.69 |
| 2 | 70.2 | 0.28 | 0.72 |
| 3 | 69.6 | 0.33 | 0.71 |

Each walker explores parameter space independently.

---

## Step 4: Evaluate the Prior

Before computing the likelihood, each proposed parameter set is checked against the prior.

Examples of prior constraints:

- $H_0 > 0$
- $0 < \Omega_M < 1.5$
- $0 \le \Omega_\Lambda < 1.5$

If a proposed point violates the prior, its posterior probability is set to zero by returning

$$
\log P = -\infty,
$$

and that point is rejected immediately.

---

## Step 5: Compute the Model Prediction

For each proposed cosmology, the predicted luminosity distance is calculated using the ΛCDM model,

$$
D_L(z),
$$

which is converted into the predicted distance modulus,

$$
\mu_{\rm model} = 5\log_{10}\left(\frac{D_L}{\rm Mpc}\right)+25.
$$

This produces one predicted distance modulus for every observed supernova.

---

## Step 6: Compare the Model to the Data

For the two- and three-parameter models,

$$ 
{\rm residual} = \mu_{\rm obs} - \mu_{\rm model},
$$

where $\mu_{\rm obs}$ is the measured distance modulus.

For the six-parameter model,

$$
\mu_{\rm obs} = m_B - M_B + \alpha x_1 - \beta c + \gamma HM,
$$

is first reconstructed from the light-curve parameters before computing

$$
{\rm residual} = \mu_{\rm obs} - \mu_{\rm model}.
$$

Small residuals indicate that the proposed cosmology and parameters reproduce the observations well.

---

## Step 7: Compute the Likelihood

Assuming Gaussian measurement uncertainties, the log-likelihood is

$$
\log L = -\frac12 \sum_i \left[ \ln(2\pi\sigma_i^2) + \frac{(\mu_{{\rm obs},i}-\mu_{{\rm model},i})^2} {\sigma_i^2} \right].
$$

Parameter combinations that produce smaller residuals have larger likelihoods.

---

## Step 8: Compute the Posterior

Bayes' theorem states

$$
P(\theta|D) \propto P(D|\theta) P(\theta),
$$

where

- $P(\theta|D)$ is the posterior,
- $P(D|\theta)$ is the likelihood,
- $P(\theta)$ is the prior.

The MCMC evaluates the logarithm of this expression:

$$
\log P = \log(\mathrm{prior}) + \log(\mathrm{likelihood}).
$$

---

## Step 9: Propose a New Position

Each walker proposes a nearby point in parameter space.

Example:

Current position

$$ 
(H_0,\Omega_M,\Omega_\Lambda) = (70.0,0.30,0.70)
$$

Proposed position
$$
(H_0,\Omega_M,\Omega_\Lambda) = (69.9,0.31,0.69).
$$

---

## Step 10: Accept or Reject the Proposal

The new point is evaluated.

- If it fits the data better, it is usually accepted.
- If it fits the data worse, it may still be accepted with some probability.

Allowing occasional worse steps prevents the walkers from becoming trapped in local maxima and enables efficient exploration of the posterior distribution.

---

## Step 11: Repeat the Process

Each walker repeatedly

1. proposes a new position,
2. evaluates the posterior,
3. accepts or rejects the proposal,

until the specified number of steps has been completed.

---

## Step 12: Remove Burn-in

The earliest part of every chain depends on the chosen starting positions.

These initial samples are discarded as the **burn-in** phase, leaving only samples that represent the equilibrium posterior distribution.

---

## Step 13: Combine All Walkers

After removing burn-in, the chains from every walker are combined into one large array.

For example,

```
H0      Omega_M   Omega_Lambda
69.92   0.301     0.712
70.08   0.294     0.701
69.85   0.316     0.725
...
```

Each row represents one plausible cosmological model sampled from the posterior distribution.

---

## Step 14: Estimate the Cosmological Parameters

For each parameter, the code computes

- the 16th percentile,
- the 50th percentile (median),
- the 84th percentile.

For example,

| Percentile | $\Omega_M$ |
|------------|-------------:|
|16%|0.27|
|50%|0.30|
|84%|0.34|

The reported result is therefore

$$ 
\Omega_M = 0.30^{+0.04}_{-0.03}.
$$

The median represents the central estimate of the posterior distribution, while the 16th and 84th percentiles define the approximate 68% credible interval.

### Covariance Matrix

The light-curve fitter simultaneously determines three light-curve parameters:

* $x_0$, which is converted into the apparent magnitude (m_B),
* $x_1$, the stretch parameter,
* $c$, the color parameter.

Because these parameters are fitted simultaneously, their uncertainties are generally correlated. These correlations are described by the covariance matrix

$$
\begin{pmatrix}
\sigma_{x_0}^2 &
\mathrm{Cov}(x_0,x_1) &
\mathrm{Cov}(x_0,c) \\
\mathrm{Cov}(x_1,x_0) &
\sigma_{x_1}^2 &
\mathrm{Cov}(x_1,c) \\
\mathrm{Cov}(c,x_0) &
\mathrm{Cov}(c,x_1) &
\sigma_c^2
\end{pmatrix}.
$$

The diagonal elements are the variances of the fitted parameters, while the off-diagonal elements are the covariances, which quantify how the uncertainties in two parameters are correlated.

The metadata file stores the off-diagonal covariance terms as the columns

* `COV_x1_x0`,
* `COV_c_x0`,
* `COV_x1_c`.

Since the cosmological analysis uses the apparent magnitude $m_B$ rather than the normalization parameter $x_0$, the covariance terms involving $x_0$ are transformed using

$$
m_B = -2.5\log_{10}(x_0) + \mathrm{constant},
$$

whose derivative is

$$
\frac{dm_B}{dx_0} = -\frac{2.5}{x_0\ln10}.
$$

Applying standard uncertainty propagation gives

$$
\mathrm{Cov}(m_B,x_1) = \frac{dm_B}{dx_0} \mathrm{Cov}(x_0,x_1),
$$

and

$$
\mathrm{Cov}(m_B,c) = \frac{dm_B}{dx_0} \mathrm{Cov}(x_0,c).
$$

These transformed covariance terms are then used when propagating the uncertainty in the reconstructed distance modulus,

$$
\mu_{\rm obs} = m_B - M_B + \alpha x_1 - \beta c + \gamma HM,
$$

ensuring that the likelihood correctly accounts for the correlated uncertainties in the light-curve parameters.


````python
# FIXED
H_0 = 70.0
sigma_int = 0.10

def log_prior(theta):
    """
    Defines which parameter values the MCMC is allowed to explore.
    Returns:
        0.0      for an allowed parameter combination
        -np.inf  for a forbidden parameter combination
    Because the posterior is evaluated in logarithmic form, returning -np.inf makes the posterior probability exactly zero.
    """
    nparam = len(theta)
    
    if nparam == 2:
        H0, Omega_M = theta
        Omega_Lambda = 1.0 - Omega_M

        redshift, mu_obs, covariance_cholesky = data

    elif nparam == 3:
        H0, Omega_M, Omega_Lambda = theta

        redshift, mu_obs, covariance_cholesky = data

    elif nparam == 5:
        M_B, Omega_M, Omega_Lambda, alpha, beta = theta
        if (
            -20.5 < M_B < -18.0
            and 0.0 < Omega_M < 1.5 
            and -1.5 <= Omega_Lambda < 1.5
            and 0.0 < alpha < 0.5
            and 0.0 < beta < 6.0
        ):
            return 0.0
        else:
            return -np.inf

    elif nparam == 6:
        M_B, Omega_M, Omega_Lambda, alpha, beta, gamma = theta
        if (
            -20.5 < M_B < -18.0
            and 0.0 < Omega_M < 1.5 
            and -1.5 <= Omega_Lambda < 1.5
            and 0.0 < alpha < 0.5
            and 0.0 < beta < 6.0
            and -0.5 < gamma < 0.5
        ):
            return 0.0
        else:
            return -np.inf
    else:
        raise ValueError("Unexpected size? len(theta) should be 2, 3, 5, or 6")    


def log_likelihood(theta, data):
    """
    Compares the measured distance moduli with the distance moduli predicted by the proposed cosmological parameters.
    Theta contains the proposed cosmological parameters.
    For 2D/3D fits, data contains redshift, observed distance modulus,
    and the Cholesky factor of the corresponding STAT+SYS covariance submatrix.
    For 5D/6D fits, data contains the SALT2 measurements and uncertainties.
    """
    nparam = len(theta)

    if nparam == 2:
        H0, Omega_M = theta
        Omega_Lambda = 1.0 - Omega_M

        redshift, mu_obs, covariance_cholesky = data

    elif nparam == 3:
        H0, Omega_M, Omega_Lambda = theta

        redshift, mu_obs, covariance_cholesky = data

    elif nparam == 5:
        M_B, Omega_M, Omega_Lambda, alpha, beta = theta
        (
            redshift,
            mB,
            x1,
            c,
            host_logmass,
            mB_err,
            x1_err,
            c_err,
            cov_mB_x1,
            cov_mB_c,
            cov_x1_c,
            muerr_vpec,
            lensdmuerr
        ) = data
        mu_obs = mB - M_B + alpha*x1 - beta*c
        variance = mB_err**2 + alpha**2*x1_err**2 + beta**2*c_err**2 + 2.0*alpha*cov_mB_x1 - 2.0*beta*cov_mB_c - 2.0*alpha*beta*cov_x1_c + muerr_vpec**2 + lensdmuerr**2 + sigma_int**2

    elif nparam == 6:
        M_B, Omega_M, Omega_Lambda, alpha, beta, gamma = theta
        (
            redshift,
            mB,
            x1,
            c,
            host_logmass,
            mB_err,
            x1_err,
            c_err,
            cov_mB_x1,
            cov_mB_c,
            cov_x1_c,
            muerr_vpec,
            lensdmuerr
        ) = data
        HM = (host_logmass >= mass_cut).astype(float)
        mu_obs = mB - M_B + alpha*x1 - beta*c + gamma*HM
        variance = mB_err**2 + alpha**2*x1_err**2 + beta**2*c_err**2 + 2.0*alpha*cov_mB_x1 - 2.0*beta*cov_mB_c - 2.0*alpha*beta*cov_x1_c + muerr_vpec**2 + lensdmuerr**2 + sigma_int**2
    else:
        raise ValueError("Unexpected size? len(theta) should be 2, 3, 5, or 6")    


    # Construct Astropy LambdaCDM cosmology using the current proposed values of H0, Omega_M, and Omega_Lambda.
    # H0: Present-day Hubble expansion rate in km/s/Mpc.
    # Om0: Present-day matter density parameter Omega_M.
    # Ode0: Present-day dark-energy density parameter Omega_Lambda.

    # Use fitted H0 for the 2D/3D models and fixed H_0 for the 5D/6D models.
    if nparam in [2, 3]:
        H0_model = H0
    else:
        H0_model = H_0
    
    cosmology = LambdaCDM(
        H0=H0_model,
        Om0=Omega_M,
        Ode0=Omega_Lambda
    )
    
    with np.errstate(invalid="ignore"):
        mu_pred = cosmology.distmod(redshift).value
    
    if not np.all(np.isfinite(mu_pred)):
        return -np.inf
    
    residuals = mu_obs - mu_pred

    if nparam in (2, 3):
        # Full correlated Pantheon+SH0ES Gaussian likelihood.
        # If C = L L^T, then chi^2 = ||L^{-1} residuals||^2.
        try:
            whitened_residuals = np.linalg.solve(covariance_cholesky, residuals)
        except np.linalg.LinAlgError:
            return -np.inf

        chi2 = whitened_residuals @ whitened_residuals
        logdet_covariance = 2.0 * np.sum(np.log(np.diag(covariance_cholesky)))
        loglike = -0.5 * (chi2 + logdet_covariance + len(residuals)*np.log(2.0*np.pi))

    else:
        if np.any(~np.isfinite(variance)) or np.any(variance <= 0.0):
            return -np.inf

        loglike = -0.5 * np.sum(np.log(2.0*np.pi*variance) + residuals**2/variance)

    return loglike


def log_posterior(theta, data):
    logpr = log_prior(theta)
    if logpr == -np.inf:
        return logpr
    else:
        return logpr + log_likelihood(theta, data)

def make_data_tuple(sample, nparam):
    if nparam in (2, 3):
        rows = sample["_PANTHEON_ROW"].to_numpy(dtype=int)
        covariance = pantheon_cov[np.ix_(rows, rows)]

        try:
            covariance_cholesky = np.linalg.cholesky(covariance)
        except np.linalg.LinAlgError as exc:
            raise ValueError("Pantheon+SH0ES covariance submatrix is not positive definite") from exc

        return (
            sample["zHD"].to_numpy(),
            sample["MU"].to_numpy(),
            covariance_cholesky
        )

    elif nparam in (5, 6):
        return (
            sample["zHD"].to_numpy(),
            sample["mB"].to_numpy(),
            sample["x1"].to_numpy(),
            sample["c"].to_numpy(),
            sample["HOST_LOGMASS"].to_numpy(),
            sample["mBERR"].to_numpy(),
            sample["x1ERR"].to_numpy(),
            sample["cERR"].to_numpy(),
            sample["COV_mB_x1"].to_numpy(),
            sample["COV_mB_c"].to_numpy(),
            sample["COV_x1_c"].to_numpy(),
            sample["MUERR_VPEC"].to_numpy(),
            sample["LENSDMUERR"].to_numpy()
        )

    else:
        raise ValueError("nparam should be 2, 3, 5, or 6")


def make_starting_positions(initial_guess, nwalkers, random_seed=None):
    rng = np.random.default_rng(random_seed)
    ndim = len(initial_guess)

    if ndim == 2:
        scales = np.array([1.0, 0.02])

    elif ndim == 3:
        scales = np.array([1.0, 0.02, 0.02])

    elif ndim == 5:
        scales = np.array([0.02, 0.02, 0.02, 0.01, 0.10])
    
    elif ndim == 6:
        scales = np.array([0.02, 0.02, 0.02, 0.01, 0.10, 0.01])

    else:
        raise ValueError("initial_guess must contain 2, 3, 5, or 6 parameters")

    positions = initial_guess + scales * rng.normal(size=(nwalkers, ndim))

    for i in range(nwalkers):
        while not np.isfinite(log_prior(positions[i])):
            positions[i] = initial_guess + scales * rng.normal(size=ndim)

    return positions


def run_mcmc(data, initial_guess, nwalkers=32, nsteps=5000, burnin=1000, thin=1, random_seed=None):
    ndim = len(initial_guess)

    if nwalkers < 2 * ndim:
        raise ValueError("nwalkers must be at least twice ndim")

    if burnin >= nsteps:
        raise ValueError("burnin must be smaller than nsteps")

    starting_positions = make_starting_positions(
        initial_guess,
        nwalkers,
        random_seed=random_seed
    )

    sampler = emcee.EnsembleSampler(
        nwalkers,
        ndim,
        log_posterior,
        args=(data,)
    )

    sampler.run_mcmc(
        starting_positions,
        nsteps,
        progress=True
    )

    samples = sampler.get_chain(
        discard=burnin,
        thin=thin,
        flat=True
    )

    return sampler, samples


def summarize_samples(name, samples, filename=None):
    if samples.shape[1] == 2:
        labels = ["H_0", "Omega_M"]
    elif samples.shape[1] == 3:
        labels = ["H_0", "Omega_M", "Omega_Lambda"]
    elif samples.shape[1] == 5:
        labels = ["M_B", "Omega_M", "Omega_Lambda", "alpha", "beta"]
    elif samples.shape[1] == 6:
        labels = ["M_B", "Omega_M", "Omega_Lambda", "alpha", "beta", "gamma"]
    else:
        raise ValueError("Samples must contain 2, 3, 5, or 6 columns")

    percentiles = np.percentile(samples, [16, 50, 84], axis=0)

    print(f"\n{name}")

    summary = {}
    summary_lines = [name]

    for i, label in enumerate(labels):
        median = percentiles[1, i]
        lower = median - percentiles[0, i]
        upper = percentiles[2, i] - median

        summary[label] = {
            "median": median,
            "minus": lower,
            "plus": upper
        }

        line = f"{label} = {median:.4f} -{lower:.4f} +{upper:.4f}"

        print(line)
        summary_lines.append(line)

    if filename is not None:
        filename = Path(filename)
        filename.parent.mkdir(parents=True, exist_ok=True)

        summary_text = "\n".join(summary_lines)

        with open(filename, "w") as f:
            f.write(summary_text)

    return summary

def get_corner_ranges(samples1, samples2):
    combined = np.vstack([samples1, samples2])
    ranges = []

    for i in range(combined.shape[1]):
        low = np.min(combined[:, i])
        high = np.max(combined[:, i])
        padding = 0.05 * (high - low)

        ranges.append((low - padding, high + padding))

    return ranges

def plot_overlaid_corner(samples_1, samples_2, label_1, label_2, title, color_1, color_2, filename=None):
    
    ndim = samples_1.shape[1]

    if samples_2.shape[1] != ndim:
        raise ValueError("Both sample arrays must have the same number of columns")

    corner_ranges = get_corner_ranges(samples_1, samples_2)
    
    if ndim == 2:
        parameter_labels = [
            r"$H_0$",
            r"$\Omega_M$"
        ]

    elif ndim == 3:
        parameter_labels = [
            r"$H_0$",
            r"$\Omega_M$",
            r"$\Omega_\Lambda$"
        ]

    elif ndim == 5:
        parameter_labels = [
            r"$M_B$",
            r"$\Omega_M$",
            r"$\Omega_\Lambda$",
            r"$\alpha$",
            r"$\beta$",
        ]
    
    elif ndim == 6:
        parameter_labels = [
            r"$M_B$",
            r"$\Omega_M$",
            r"$\Omega_\Lambda$",
            r"$\alpha$",
            r"$\beta$",
            r"$\gamma$"
        ]

    else:
        raise ValueError("Samples must contain 2, 3, 5, or 6 columns")
    
    figure = corner.corner(
        samples_1,
        labels=parameter_labels,
        range=corner_ranges,
        color=color_1,
        levels=(0.683, 0.954),
        plot_datapoints=False,
        plot_density=True,
        fill_contours=False,
        smooth=1.0,
        show_titles=False,
        title_fmt=".3f"
    )

    corner.corner(
        samples_2,
        labels=parameter_labels,
        range=corner_ranges,
        color=color_2,
        levels=(0.683, 0.954),
        plot_datapoints=False,
        plot_density=True,
        fill_contours=False,
        smooth=1.0,
        show_titles=False,
        fig=figure
    )

    legend_handles = [
        Line2D([0], [0], color=color_1, linewidth=2, label=label_1),
        Line2D([0], [0], color=color_2, linewidth=2, label=label_2)
    ]

    figure.legend(
        handles=legend_handles,
        loc="upper right",
        frameon=False
    )

    figure.suptitle(title, fontsize=14, y=1.02)
    #figure.subplots_adjust(top=0.90)

    if filename is not None:
        figure.savefig(
            filename,
            dpi=300,
            bbox_inches="tight"
        )   
    plt.show()

# DIAGNOSTICS
def plot_x1_diagnostic(sample, M_B, Omega_M, Omega_Lambda, alpha, beta, gamma, title=r"$x_1$ diagnostic", filename=None):
    data = sample[["zHD", "mB", "x1", "c", "HOST_LOGMASS"]].dropna().copy()

    cosmology = LambdaCDM(H0=H_0, Om0=Omega_M, Ode0=Omega_Lambda)
    data["MU_MODEL"] = cosmology.distmod(data["zHD"].to_numpy()).value

    HM = (data["HOST_LOGMASS"] >= mass_cut).astype(float)

    data["y"] = data["mB"] - M_B - data["MU_MODEL"] - beta*data["c"] + gamma*HM

    plt.figure(figsize=(7, 5))
    plt.scatter(data["x1"], data["y"], s=15, alpha=0.4)

    slope, intercept = np.polyfit(data["x1"], data["y"], 1)
    x1_line = np.linspace(data["x1"].min(), data["x1"].max(), 100)

    plt.plot(x1_line, slope*x1_line + intercept)
    plt.xlabel(r"$x_1$")
    plt.ylabel(r"$m_B-M_B-\mu_{\rm model}-\beta c+\gamma HM$")
    plt.title(title)
    plt.text(
        0.05, 0.95,
        f"Measured slope = {slope:.4f}\nExpected slope = {-alpha:.4f}",
        transform=plt.gca().transAxes,
        verticalalignment="top",
        bbox=dict(facecolor="white", alpha=0.8)
    )
    plt.tight_layout()
    if filename is not None:
        plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Measured slope = {slope:.4f}")
    print(f"Expected slope = {-alpha:.4f}")

def plot_c_diagnostic(sample, M_B, Omega_M, Omega_Lambda, alpha, beta, gamma, title=r"$c$ diagnostic", filename=None):
    data = sample[["zHD", "mB", "x1", "c", "HOST_LOGMASS"]].dropna().copy()

    cosmology = LambdaCDM(H0=H_0, Om0=Omega_M, Ode0=Omega_Lambda)
    data["MU_MODEL"] = cosmology.distmod(data["zHD"].to_numpy()).value

    HM = (data["HOST_LOGMASS"] >= mass_cut).astype(float)

    data["y"] = data["mB"] - M_B - data["MU_MODEL"] + alpha*data["x1"] + gamma*HM

    plt.figure(figsize=(7, 5))
    plt.scatter(data["c"], data["y"], s=15, alpha=0.4)

    slope, intercept = np.polyfit(data["c"], data["y"], 1)
    c_line = np.linspace(data["c"].min(), data["c"].max(), 100)

    plt.plot(c_line, slope*c_line + intercept)
    plt.xlabel(r"$c$")
    plt.ylabel(r"$m_B-M_B-\mu_{\rm model}+\alpha x_1+\gamma HM$")
    plt.title(title)
    plt.text(
        0.05, 0.95,
        f"Measured slope = {slope:.4f}\nExpected slope = {beta:.4f}",
        transform=plt.gca().transAxes,
        verticalalignment="top",
        bbox=dict(facecolor="white", alpha=0.8)
    )
    plt.tight_layout()
    if filename is not None:
        plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Measured slope = {slope:.4f}")
    print(f"Expected slope = {beta:.4f}")

def plot_gamma_diagnostic(sample, M_B, Omega_M, Omega_Lambda, alpha, beta, gamma, title=r"$\gamma$ diagnostic", filename=None):
    data = sample[["zHD", "mB", "x1", "c", "HOST_LOGMASS"]].dropna().copy()

    cosmology = LambdaCDM(H0=H_0, Om0=Omega_M, Ode0=Omega_Lambda)
    data["MU_MODEL"] = cosmology.distmod(data["zHD"].to_numpy()).value

    HM = (data["HOST_LOGMASS"] >= mass_cut).astype(float)

    data["y"] = data["mB"] - M_B - data["MU_MODEL"] + alpha*data["x1"] - beta*data["c"]

    low = data["y"][HM == 0]
    high = data["y"][HM == 1]

    low_mean = low.mean()
    high_mean = high.mean()
    measured_step = high_mean - low_mean

    plt.figure(figsize=(7, 5))
    plt.scatter(data["HOST_LOGMASS"], data["y"], s=15, alpha=0.4)

    plt.hlines(low_mean, data["HOST_LOGMASS"].min(), mass_cut, linewidth=2)
    plt.hlines(high_mean, mass_cut, data["HOST_LOGMASS"].max(), linewidth=2)
    plt.axvline(mass_cut, linestyle="--")

    plt.xlabel(r"Host Stellar Mass, $\log(M_*/M_\odot)$")
    plt.ylabel(r"$m_B-M_B-\mu_{\rm model}+\alpha x_1-\beta c$")
    plt.title(title)
    plt.text(
        0.05, 0.95,
        f"Measured step = {measured_step:.4f}\nExpected step = {-gamma:.4f}",
        transform=plt.gca().transAxes,
        verticalalignment="top",
        bbox=dict(facecolor="white", alpha=0.8)
    )
    plt.tight_layout()
    if filename is not None:
        plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Measured high-mass - low-mass step = {measured_step:.4f}")
    print(f"Expected step = {-gamma:.4f}")

def plot_mcmc_diagnostics(sample, samples, title="MCMC diagnostics", filename_prefix="Pantheon+SH0ES", clipped=False):
    ndim = samples.shape[1]

    if clipped:
        if ndim == 5:
            diagnostics_dir = clipped_5d_dir
        elif ndim == 6:
            diagnostics_dir = clipped_6d_dir
        else:
            raise ValueError("Diagnostics only supported for 5D or 6D samples")
    else:
        if ndim == 5:
            diagnostics_dir = nonclipped_5d_dir
        elif ndim == 6:
            diagnostics_dir = nonclipped_6d_dir
        else:
            raise ValueError("Diagnostics only supported for 5D or 6D samples")

    M_B_fit = np.median(samples[:, 0])
    Omega_M_fit = np.median(samples[:, 1])
    Omega_Lambda_fit = np.median(samples[:, 2])
    alpha_fit = np.median(samples[:, 3])
    beta_fit = np.median(samples[:, 4])

    if ndim == 5:
        gamma_fit = 0.0
    else:
        gamma_fit = np.median(samples[:, 5])

    plot_x1_diagnostic(
        sample,
        M_B_fit,
        Omega_M_fit,
        Omega_Lambda_fit,
        alpha_fit,
        beta_fit,
        gamma_fit,
        title=rf"$x_1$ diagnostic: {title}",
        filename=diagnostics_dir / f"{filename_prefix}_x1_diagnostic.png"
    )

    plot_c_diagnostic(
        sample,
        M_B_fit,
        Omega_M_fit,
        Omega_Lambda_fit,
        alpha_fit,
        beta_fit,
        gamma_fit,
        title=rf"$c$ diagnostic: {title}",
        filename=diagnostics_dir / f"{filename_prefix}_c_diagnostic.png"
    )

    if ndim == 6:
        plot_gamma_diagnostic(
            sample,
            M_B_fit,
            Omega_M_fit,
            Omega_Lambda_fit,
            alpha_fit,
            beta_fit,
            gamma_fit,
            title=rf"$\gamma$ diagnostic: {title}",
            filename=diagnostics_dir / f"{filename_prefix}_gamma_diagnostic.png"
        )
````

````python
required_columns = [
    "_PANTHEON_ROW",
    "zHD",
    "MU",
    "MUERR",
    "HOST_LOGMASS",
    "mB",
    "x1",
    "c",
    "mBERR",
    "x1ERR",
    "cERR",
    "COV_mB_x1",
    "COV_mB_c",
    "COV_x1_c",
    "MUERR_VPEC",
    "LENSDMUERR"
]

pantheon_sample = metadata[required_columns].replace([np.inf, -np.inf], np.nan).dropna().copy()
pantheon_sample = pantheon_sample[(pantheon_sample["zHD"] > 0) & (pantheon_sample["MUERR"] > 0)].copy()

# Remove rows with invalid SALT2 covariance information for the 5D/6D fits.
rho_mB_x1 = pantheon_sample["COV_mB_x1"] / (pantheon_sample["mBERR"] * pantheon_sample["x1ERR"])
rho_mB_c = pantheon_sample["COV_mB_c"] / (pantheon_sample["mBERR"] * pantheon_sample["cERR"])
rho_x1_c = pantheon_sample["COV_x1_c"] / (pantheon_sample["x1ERR"] * pantheon_sample["cERR"])
valid_covariance = np.isfinite(rho_mB_x1) & np.isfinite(rho_mB_c) & np.isfinite(rho_x1_c) & (np.abs(rho_mB_x1) <= 1.0) & (np.abs(rho_mB_c) <= 1.0) & (np.abs(rho_x1_c) <= 1.0)

pantheon_sample_6d = pantheon_sample[valid_covariance].copy()
pantheon_sample_6d = pantheon_sample_6d[(pantheon_sample_6d["mBERR"] > 0) & (pantheon_sample_6d["x1ERR"] > 0) & (pantheon_sample_6d["cERR"] > 0)].copy()
pantheon_sample_5d = pantheon_sample_6d.copy()

mass_cut = 10.0
print(f"HOST_LOGMASS cut = {mass_cut:.1f}")
print(f"Original sample size = {len(pantheon_sample)}")
print(f"Covariance-cleaned 5D/6D sample = {len(pantheon_sample_6d)}")
````

````python
def plot_hubble_split(
    sample1,
    sample2,
    samples1,
    samples2,
    label1,
    label2,
    color1,
    color2,
    title,
    filename
):
    M_B_1 = np.median(samples1[:, 0])
    Omega_M_1 = np.median(samples1[:, 1])
    Omega_Lambda_1 = np.median(samples1[:, 2])
    alpha_1 = np.median(samples1[:, 3])
    beta_1 = np.median(samples1[:, 4])
    gamma_1 = np.median(samples1[:, 5]) if samples1.shape[1] > 5 else 0.0

    M_B_2 = np.median(samples2[:, 0])
    Omega_M_2 = np.median(samples2[:, 1])
    Omega_Lambda_2 = np.median(samples2[:, 2])
    alpha_2 = np.median(samples2[:, 3])
    beta_2 = np.median(samples2[:, 4])
    gamma_2 = np.median(samples2[:, 5]) if samples2.shape[1] > 5 else 0.0

    data1 = sample1[["zHD", "mB", "x1", "c", "HOST_LOGMASS"]].dropna().copy()
    data2 = sample2[["zHD", "mB", "x1", "c", "HOST_LOGMASS"]].dropna().copy()

    HM1 = (data1["HOST_LOGMASS"] >= mass_cut).astype(float)
    HM2 = (data2["HOST_LOGMASS"] >= mass_cut).astype(float)

    data1["MU"] = data1["mB"] - M_B_1 + alpha_1*data1["x1"] - beta_1*data1["c"] + gamma_1*HM1
    data2["MU"] = data2["mB"] - M_B_2 + alpha_2*data2["x1"] - beta_2*data2["c"] + gamma_2*HM2

    cosmology1 = LambdaCDM(H0=H_0, Om0=Omega_M_1, Ode0=Omega_Lambda_1)
    cosmology2 = LambdaCDM(H0=H_0, Om0=Omega_M_2, Ode0=Omega_Lambda_2)

    z_min = min(data1["zHD"].min(), data2["zHD"].min())
    z_max = max(data1["zHD"].max(), data2["zHD"].max())
    z_plot = np.linspace(z_min, z_max, 500)

    plt.figure(figsize=(7, 5))

    plt.scatter(data1["zHD"], data1["MU"], s=15, alpha=0.4, color=color1, label=label1)
    plt.scatter(data2["zHD"], data2["MU"], s=15, alpha=0.4, color=color2, label=label2)

    plt.plot(z_plot, cosmology1.distmod(z_plot).value, color=color1, linewidth=2)
    plt.plot(z_plot, cosmology2.distmod(z_plot).value, color=color2, linewidth=2)

    plt.xscale("log")
    plt.xlabel("Redshift, z")
    plt.ylabel(r"Distance modulus, $\mu$")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()
````

This DES-specific color/sSFR or clipped-analysis cell is not run in the Pantheon+SH0ES version because the required host columns are absent or because the Pantheon+SH0ES release requires covariance-aware clipping/fitting.
